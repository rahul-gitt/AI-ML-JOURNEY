```python
# A researcher wants to compare the average exam scores of Class A and Class B.

import numpy as np
from scipy import stats

class_A = [
72, 75, 78, 80, 74,
76, 79, 81, 77, 73,
75, 78, 80, 76, 74
]

class_B = [
68, 70, 72, 69, 71,
73, 70, 72, 69, 71,
68, 70, 73, 72, 69
]
```

```python
t_stats, p_value = stats.ttest_ind(class_A,class_B, equal_var=False)
t_stats
```

```python
alpha = 0.05
if p_value < alpha:
    print("I will reject the null hypothesis.")
else:
    print("I will accept the null hypothesis.")
```
