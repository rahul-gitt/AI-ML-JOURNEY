Linear Regression


```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

X = [[1], [2], [3], [4], [5]]
y = [27, 40, 60, 70, 80]

model.fit(X,y)

hours = float(input("Enter your study Hour : "))
marks = model.predict([[hours]])

print(f"Your marks is {marks}")
```
