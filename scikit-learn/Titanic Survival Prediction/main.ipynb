```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")
```

```python
df = sns.load_dataset("titanic")
df.head()
```

```python
df.drop('deck', axis = 1, inplace= True)
```

```python
df.drop('who', axis= 1, inplace = True)
```

```python
df.drop(["adult_male", "class", 'alive', 'embark_town'],
        axis = 1,
        inplace= True)
```

```python
df.info()
```

```python
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df.dropna(subset= ['embarked'], inplace= True)
```

```python
df.info()
```

```python
df
```

```python
df['sex'].value_counts()
```

```python
df['sex'] = df['sex'].map({
    'male' : 1,
    'female' : 0
})
```

```python
df['embarked'].value_counts()
```

```python
df = pd.get_dummies(
    df,
    columns = ['embarked'],
    drop_first= True,
    dtype= int
)
```

```python
df['alone'] = df['alone'].astype(int)
```
