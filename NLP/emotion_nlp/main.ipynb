```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.txt',sep = ';', header= None, names = ['text', 'emotions'])
```

```python
unique_emotions = df['emotions'].unique()
emotions_num = {}
```

Numbering all emotions 

```python
i = 0 
for emo in unique_emotions :
    emotions_num[emo] = i
    i+= 1

df['emotions'] = df['emotions'].map(emotions_num)
```

```python
df
```

Convert everything in lower case

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

Remove Punctuations

```python
import string

def remove_punc(txt):
    return txt.translate(str.maketrans('', '',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

remove digits

```python
def remove_digit(txt):
    new = ''
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_digit)
```

Remove URLs/ Link

```python
df['text'] = df['text'].str.replace(r'http\S+|www\S+', '', regex=True)
```

remove html tags

```python
df['text'] = df['text'].str.replace(r'<.*?>', '', regex=True)
```

remove emojis & special char

```python
def remove_emoji(txt):
    new = ''
    for i in txt:
        if i.isascii():
            new += i
        return new

df['text'] = df['text'].apply(remove_emoji)
```

remove stop words

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stopwords = stopwords.words('english')
```

```python
def remove(txt):
    words = word_tokenize(txt)
    cleaned = []
    for i in words:
        if not i in stopwords:
            cleaned.append(i)

    return ' '.join(cleaned)
```
