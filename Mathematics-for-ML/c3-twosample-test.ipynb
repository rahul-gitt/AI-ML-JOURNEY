```python
# A researcher wants to compare the average exam scores of Class A and Class B.

import numpy as np
from scipy import stats

class_A = [
72, 75, 78, 80, 74,
76, 79, 81, 77, 73,
75, 78, 80, 76, 74
]

class_B = [
68, 70, 72, 69, 71,
73, 70, 72, 69, 71,
68, 70, 73, 72, 69
]
```

```python
t_stats, p_value = stats.ttest_ind(class_A,class_B, equal_var=False)
t_stats, p_value
```

```python
alpha = 0.05
if p_value < alpha:
    print("I will reject the null hypothesis.")
else:
    print("I will accept the null hypothesis.")
```

```python
# A pharmaceutical company wants to compare the average 
# recovery time (in days) of patients using Medicine A and Medicine B.

from scipy import stats
import numpy as np

medicine_A = [
12, 11, 13, 14, 12,
15, 13, 14, 12, 11,
13, 12, 14, 15, 13,
12, 14, 13, 11, 12
]

medicine_B = [
15, 16, 14, 17, 15,
16, 15, 17, 14, 16,
15, 16, 17, 15, 14,
16, 15, 17, 16, 15
]
alpha = 0.05
```

```python
t_stats, p_value = stats.ttest_ind(medicine_A,medicine_B, equal_var=False)
t_stats
```

```python
if t_stats < alpha :
    print("I will reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
