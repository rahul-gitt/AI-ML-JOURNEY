Linear Regression


```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

X = [[1], [2], [3], [4], [5]]
y = [27, 40, 60, 70, 80]

model.fit(X,y)

hours = float(input("Enter your study Hour : "))
marks = model.predict([[hours]])

print(f"Your marks is {marks}")
```

Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()

X = [[1], [2], [3], [4], [5]]
y = [0,0,1,1,1]

model.fit(X,y)
hours = float(input('Enter your study hour : '))

result = model.predict([[hours]])

if result == 1:
    print(f"Bassed on your study Hour {hours} you are likely Pass")
else:
    print(f"Bassed on your study hour {hours} you are likely fail")
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=3)


X = [[180, 7],
     [200,7.5],
     [250, 8],
     [300, 8.5],
     [330, 9],
     [360, 9.5]]

y = [0,0,0,1,1,1]

model.fit(X,y)

weight = float(input("Enter weight in grams : "))
size = float(input('Enter size in cm : '))

result= model.predict([[weight, size]])[0]

if result == 0:
    print(f"Bassed on weight {weight}, size {size} your fruit is likely Apple")

else:
    print(f"Bassed on weight {weight}, size {size} your fruit is likely Orange")
```
