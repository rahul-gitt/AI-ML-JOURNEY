```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.txt', sep=';', header= None, names= ['text', 'emotions'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emo = df['emotions'].unique()
emotions_num = {}
```

```python
i = 0
for emo in unique_emo:
    emotions_num[emo] = i
    i += 1

df['emotions'] = df['emotions'].map(emotions_num)
```

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
import string

def remove_punc(txt):
    return txt.translate(str.maketrans('', '', string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

```python
def remove_num(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new += i
    return new

df['text'] = df['text'].apply(remove_num)
```

```python
def remove_emo(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i

    return new

df['text'] = df['text'].apply(remove_emo)
```

Remove Stop Words

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
stop_words = set(stopwords.words('english'))
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
    words = word_tokenize(txt)
    cleaned = []
    for i in words:
        if not i in stop_words:
            cleaned.append(i)

    return ' '. join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```
