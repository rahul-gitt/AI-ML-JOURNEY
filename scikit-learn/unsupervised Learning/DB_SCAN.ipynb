Learn DB_SCAN Clustering

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_moons
```

```python
X, y = make_moons(n_samples= 500, noise = 0.05, random_state= 42)
```

K means implementation

```python
from sklearn.cluster import KMeans, DBSCAN
```

```python
df = pd.DataFrame(X, columns= ['Fiture1', 'Fiture2'])
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
kmeans = KMeans(n_clusters= 2, random_state= 42)
kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python
df['kmeans_cluster'] = kmeans_labels
```

```python
sns.scatterplot(x = df['Fiture1'], y= df['Fiture2'], hue= df['kmeans_cluster'],
                )
```

DB SCAN implementation

```python
dbscan = DBSCAN(eps = 0.3, min_samples= 5)
dbscan_labels = dbscan.fit_predict(X_scaled)
```

```python
df['dbscan_cluster'] = dbscan_labels
```

```python
sns.scatterplot(x = df['Fiture1'], y= df['Fiture2'], hue= df['dbscan_cluster'],
                )
```
