```python
import numpy as np
from scipy.stats import norm 
```

```python
sample = [
68,72,75,70,69,74,73,71,76,67,
72,74,70,71,73,69,68,75,72,71,
70,74,73,72,69,68,71,75,74,70,
73,72,71,69,68,74,75,70,72,73,
71,69,68,74,75,72,70,73,71,69,
68,74,75,70,71,72,73,69,68,74,
75,71,70,72,73,69,68,74,75,70,
71,72,73,69,68,74,75,70,71,72,          # sample data
73,69,68,74,75,70,71,72,73,69,
68,74,75,70,71,72,73,69,68,74
]

population_mean = 70
population_std = 4.5
sample_mean = np.mean(sample)

n = len(sample)   # len of the data set
```

```python
z_score = (sample_mean - population_mean)/ (population_std/ np.sqrt(n))
z_score
```

```python
alpha = 0.05
p_value = 2 * (1 - norm.cdf(abs(z_score)))
p_value
```

```python
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print('Accept the null hypothesis')
```
