```python
Q1 = np.percentile(arr, 25)

Q3 = np.percentile(arr, 75)

IQR = Q3 - Q1
```

```python
LF = Q1 - (1.5  * IQR)

UF = Q3 + (1.5 * IQR)
```

```python
l = []
for i in arr :
    if i >= LF and i <= UF :
        l.append(i)
arr2 = np.array(l)

print(arr2)
```
